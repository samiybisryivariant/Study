# ДЗ 2.3.10

Функции:
1. случайный список чисел из равномерного распределения между a и b
2. частотный словарь по списку
3. базовые статистики списка

```python
import random
from collections import Counter
```

## 1. Случайный список

```python
def random_values(a: int, b: int, length: int) -> list:
    result = []
    for _ in range(length):
        value = int(random.uniform(a, b))
        result.append(value)
    return result
```

```python
nums = random_values(1, 10, 20)
nums
```

## 2. Частотный словарь

Пример: `[1, 2, 3, 1, 2, 2]` -> `{1: 2, 2: 3, 3: 1}`

```python
def value_counts(values: list) -> dict:
    return dict(Counter(values))
```

```python
value_counts([1, 2, 3, 1, 2, 2])
```

```python
value_counts(nums)
```

## 3. Статистики списка

```python
def list_stats(values: list) -> dict:
    ordered = sorted(values)
    n = len(ordered)
    mid = n // 2
    if n % 2 == 0:
        median_value = (ordered[mid - 1] + ordered[mid]) / 2
    else:
        median_value = ordered[mid]
    mode_value = Counter(values).most_common(1)[0][0]
    return {
        "min": ordered[0],
        "mean": sum(values) / n,
        "median": median_value,
        "mode": mode_value,
        "max": ordered[-1],
    }
```

```python
list_stats(nums)
```
